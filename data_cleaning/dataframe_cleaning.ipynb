# Cleaning Pandas DataFrames

```python
from pathlib import Path
import pandas as pd
```

## DataFrame Inspection

```python
data_folder = Path('../../data/')
old_path = Path(f'{data_folder}/nederland_pop_stats_old.csv')
```

```python
pop_stats = pd.read_csv(old_path)
pop_stats.shape
```

```python
pop_stats.head(1)
```

## Deleting Unwanted Columns

```python
for i,v in enumerate(pop_stats.columns):
    print(i, v)
```

```python
pop_stats = pop_stats.drop(columns=pop_stats.columns[[0, 2, 9, 12, 13, 16, 17, 21]])

for i,v in enumerate(pop_stats.columns):
    print(i, v)
```

```python
pop_stats.head(1)
```

## Renaming Columns

```python
new_columns = {pop_stats.columns[i]: new_name for i, new_name in enumerate([
    'region',
    'total_population',
    'population_density',
    'total_births',
    'birth_ratio',
    'total_deaths',
    'death_ratio',
    'total_arrivals',
    'arrival_ratio',
    'total_departures',
    'departure_ratio',
    'net_migration',
    'total_population_growth',
    'population_growth_ratio'
])}

pop_stats = pop_stats.rename(columns=new_columns)
pop_stats.reset_index()
pop_stats.head(1)
```

## Replacing NaN Values

```python
pop_stats.isna().sum()
```

```python
pop_stats = pop_stats.fillna(0)
pop_stats.isna().sum()
```

```python
pop_stats.head(1)
```

## Removing Commas from Numeric Columns

```python
pop_stats.apply(lambda col: col.astype(str).str.contains(','), axis=1).sum()
```

```python
pop_stats = pop_stats.replace({',': ''}, regex=True)
```

```python
pop_stats.head(1)
```

## Typecasting Columns

```python
pop_stats.dtypes
```

```python
for column in pop_stats.columns[1:]:
    pop_stats[column] = pd.to_numeric(pop_stats[column])

pop_stats.dtypes
```

```python
pop_stats['population_density'] = pop_stats['population_density'].astype('float64')
pop_stats['population_density'].dtypes
```

## Removing Non-Provinces

```python
for i in pop_stats['region']:
    print(i)
```

```python
pop_stats = pop_stats[pop_stats['region'].astype('str').str.contains('(PV)', regex=False)]
pop_stats = pop_stats.reset_index()

pop_stats
```

```python
pop_stats.shape
```

## Removing Unwanted Suffixes

```python
pop_stats = pop_stats.replace({r' \(PV\)': ''}, regex=True)
pop_stats
```

## Exporting Final Product

```python
new_path = Path(f'{data_folder}/new_nederland_pop_stats.feather')
pop_stats.to_feather(new_path)
```
