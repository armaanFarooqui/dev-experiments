# Downloading a Single Sentinel 2 Image

```python
from pystac_client import Client
from odc.stac import load
import fiona

import geopandas as gpd
import rioxarray as rxr
import matplotlib.pyplot as plt

import pandas as pd
from collections import defaultdict
import xarray as xr
```

## Exploring Element 84's Earth Search Catalogue
### Connecting to the API

```python
client = Client.open('https://earth-search.aws.element84.com/v1')
```

### Printing Collection Names

```python
collections = client.get_collections()

for collection in collections:
    print(collection.id)
```

### Selecting the Sentinel Collection

```python
collection = client.get_collection('sentinel-2-l2a')
collection
```

## Creating the Enschede Extent Geometry

```python
data_folder = '../data'
fiona.listlayers(f'{data_folder}/vectors/nederland_admin_areas.gpkg')
```

```python
cities = gpd.read_file(f'{data_folder}/vectors/nederland_admin_areas.gpkg', layer='cities')
cities.shape
```

```python
enschede = cities[cities['naam']=='Enschede']
enschede.shape
```

```python
print(enschede.crs)
```

```python
enschede = enschede.to_crs('EPSG:4326')
enschede.head(1)
```

```python
enschede_geometry = enschede['geometry'].iloc[0]
enschede_geometry
```

## Retrieving Sentinel Data
### Running the Search Query

```python
results = client.search(
    collections=['sentinel-2-l2a'],
    intersects=enschede_geometry,
    datetime='2025-06-01/2025-08-31',
    query=['eo:cloud_cover<5'],
)

items = results.item_collection()
print(f'Found {len(items)} items')
```

### Selecting an Item

```python
for i, v in enumerate(items):
    print(f'{i}: {v.id}')
```

```python
item = items[0]
```

## Inspecting a Single Item
### Assets

```python
assets = item.assets
assets
```

### Properties

```python
item.bbox
```

### Visualising the Spatial Extent

```python
raster = gpd.GeoDataFrame.from_features([item], crs='EPSG:4326')
raster.explore()
```

## Creating a Multi-Band Raster
### Building the Raster Array

```python
red   = rxr.open_rasterio(assets['red'].href).squeeze("band", drop=True)
green = rxr.open_rasterio(assets['green'].href).squeeze("band", drop=True)
blue  = rxr.open_rasterio(assets['blue'].href).squeeze("band", drop=True)

rgb = xr.concat([red, green, blue], dim="band")
rgb = rgb.assign_coords(band=["red", "green", "blue"])
rgb.dims
```

### Plotting the Raster

```python
fig, ax = plt.subplots(figsize=(10,10))
rgb.plot.imshow(ax=ax, robust=True)

ax.set_title(f'Sentinel 2 L2A Image: {item.id} ({rgb.rio.crs})')
ax.set_xlabel('Easting (m)')
ax.set_ylabel('Northing (m)')

plt.show()
```

### Visualising the Raster Extent

```python
nederland = gpd.read_file('../data/vectors/nederland_admin_areas.gpkg', layer='country')

enschede = enschede.to_crs('EPSG:28992')
enschede_centroid = enschede.geometry.centroid

rgb = rgb.rio.reproject('EPSG:28992')
```

```python
fig, ax = plt.subplots(figsize=(10,10))

rgb.plot.imshow(ax=ax, robust=True)
nederland.plot(ax=ax, color='none')
enschede_centroid.plot(ax=ax, color='red', marker='o', markersize=12)

ax.annotate(
    xy=(enschede_centroid.x.iloc[0], enschede_centroid.y.iloc[0] - 0.75 * 10**4), 
    text='Enschede', 
    ha='center', 
    fontsize=9, 
    color='white')

ax.set_title(f"The Netherlands: Visualising the Extent of '{item.id}' Sentinel Image")
ax.set_xlabel('Easting (m)')
ax.set_ylabel('Northing (m)')

plt.show()
```

## Exporting the Raster

```python
rgb.rio.to_raster('enschede_sentinel_2_l2a.tif')
```
